```python
%pip install seaborn
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Налаштування стилю графіків
sns.set_theme(style="whitegrid")
plt.rcParams['font.sans-serif'] = 'DejaVu Sans'
```

```python
df_clean = df.copy()

# Очищення колонки Valuation ($B) від символів та конвертація в числа
if df_clean['Valuation ($B)'].dtype == object:
    df_clean['Valuation_Num'] = (
        df_clean['Valuation ($B)']
        .astype(str)
        .str.replace('$', '', regex=False)
        .str.replace('B', '', regex=False)
        .str.replace(',', '', regex=False)
        .str.strip()
        .astype(float)
    )
else:
    df_clean['Valuation_Num'] = pd.to_numeric(df_clean['Valuation ($B)'], errors='coerce')

# Конвертація дати та виділення року
df_clean['Date Joined'] = pd.to_datetime(df_clean['Date Joined'], errors='coerce')
df_clean['Join Year'] = df_clean['Date Joined'].dt.year

print("Очищення даних виконано успішно!")
```

```python
plt.figure(figsize=(12, 6))

# Підготовка даних для барплоту
yearly_counts = df_clean['Join Year'].value_counts().sort_index().reset_index()
yearly_counts.columns = ['Join Year', 'Count']

ax = sns.barplot(
    data=yearly_counts,
    x='Join Year',
    y='Count',
    palette='Blues_d'
)

plt.title('Динаміка появи компаній-єдинорогів за роками', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Рік приєднання', fontsize=12)
plt.ylabel('Кількість компаній', fontsize=12)
plt.xticks(rotation=45)

# Додаємо числові значення над стовпчиками
for p in ax.patches:
    ax.annotate(f'{int(p.get_height())}', 
                (p.get_x() + p.get_width() / 2., p.get_height()), 
                ha='center', va='bottom', 
                xytext=(0, 3), 
                textcoords='offset points')

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 6))

top_industries = df_clean['Industry'].value_counts().head(10).reset_index()
top_industries.columns = ['Industry', 'Count']

sns.barplot(
    data=top_industries,
    x='Count',
    y='Industry',
    palette='crest'
)

plt.title('Топ-10 галузей за кількістю компаній-єдинорогів', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Кількість компаній', fontsize=12)
plt.ylabel('Галузь (Industry)', fontsize=12)

plt.tight_layout()
plt.show()
```

```python
import numpy as np

df_clean = df.copy()

# Універсальне очищення колонки Valuation ($B): витягуємо всі цифри та крапки за допомогою регулярного виразу
df_clean['Valuation_Num'] = (
    df_clean['Valuation ($B)']
    .astype(str)
    .str.extract(r'([0-9]+\.?[0-9]*)')[0]
    .astype(float)
)

# Перевіряємо, чи з'явилися значення
print("Кількість непорожніх значень після очищення:", df_clean['Valuation_Num'].notna().sum())
print("Перші 5 очищених значень:")
print(df_clean['Valuation_Num'].head())

# Конвертація дати та виділення року
df_clean['Date Joined'] = pd.to_datetime(df_clean['Date Joined'], errors='coerce')
df_clean['Join Year'] = df_clean['Date Joined'].dt.year
```

```python
plt.figure(figsize=(12, 6))

top_countries = df_clean['Country'].value_counts().head(10)

ax = top_countries.plot(kind='bar', color='#ff7f0e', edgecolor='black')

plt.title('Топ-10 країн за кількістю компаній-єдинорогів', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Країна', fontsize=12)
plt.ylabel('Кількість компаній', fontsize=12)
plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Додаємо цифри над стовпчиками
for p in ax.patches:
    ax.annotate(str(int(p.get_height())), 
                (p.get_x() + p.get_width() / 2., p.get_height()), 
                ha='center', va='bottom', 
                xytext=(0, 3), 
                textcoords='offset points')

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

plt.scatter(
    df_clean['Founded Year'], 
    df_clean['Valuation_Num'], 
    color='purple', 
    alpha=0.5, 
    edgecolors='k', 
    s=50
)

# Обмежуємо вісь X для кращого масштабу (прибираємо аномальні роки, якщо такі є)
plt.xlim(1980, 2024)
plt.ylim(0, 50) # обмежуємо оцінку до 50 млрд, щоб основна масу було добре видно

plt.title('Зв\'язок між роком заснування та оцінкою компанії', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Рік заснування (Founded Year)', fontsize=12)
plt.ylabel('Оцінка ($B)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 6))

# Рахуємо сумарну оцінку для кожної галузі
top_valuation_industries = df_clean.groupby('Industry')['Valuation_Num'].sum().sort_values(ascending=False).head(10)

ax = top_valuation_industries.plot(kind='barh', color='#9467bd', edgecolor='black')
plt.gca().invert_yaxis()  # Найбільші зверху

plt.title('Топ-10 галузей за сумарною оцінкою (Total Valuation у млрд $)', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Сумарна оцінка ($B)', fontsize=12)
plt.ylabel('Галузь (Industry)', fontsize=12)
plt.grid(axis='x', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```
